```python
import subprocess
import sys
import re

# --- CONFIGURATION ---
dataset = "cora"  # Options: "cora", "arxiv", "mag"
batch_size = 32
# ---------------------

# Script is in the same directory
target_script = "train_jigsaw_model.py" 
# FIX: Use --batch-size (kebab-case) instead of --batch_size
cmd = f"modal run {target_script} --dataset {dataset} --batch-size {batch_size}"
log_filename = f"training_log_{dataset}.txt"

# Spam patterns to ignore
ignore_patterns = [
    r"Creating objects", r"Creating mount", r"Uploaded", r"Finalizing index",
    r"Creating function", r"Created objects", r"Initializing...", 
    r"Running app", r"Worker assigned", r"Loading images", 
    r"Running \(\d+/\d+ containers active\)",
    r"Created mount", r"Created function",
    r"Mounting .+"
]

# Regex for lines that start with a spinner (e.g., "/ Running...")
spinner_start = r"^[|/\\-]\s"

print(f"🚀 Launching Modal app for {dataset}...")
print(f"📋 Logging to {log_filename}")
print(f"🔧 Command: {cmd}")

with open(log_filename, "w", encoding="utf-8") as f:
    process = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    
    url_printed = False

    for line in process.stdout:
        # Check for URL
        if "View app at" in line or "modal.com" in line:
            if not url_printed:
                f.write(line)
                f.flush()
                print(f"🔗 {line.strip()}")
                url_printed = True
            continue # Skip subsequent prints of the URL

        # Check against specific ignore patterns
        if any(re.search(p, line) for p in ignore_patterns):
            continue

        # Filter out lines that are JUST a spinner and status
        if re.match(spinner_start, line):
            continue
            
        # Write everything else (progress bars, app logs)
        f.write(line)
        f.flush()

    process.wait()
print(f"✅ Training finished for {dataset}!")
```
